## Total Amount of Fertilizers used by the 50 Farmers

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel ("jpil_farmers_dataset.xlsx", header= 1)
total_fertilizer = df ["Fertilizer Used kg"].sum()
print(f"Total Fertilizer used by the 50 Farmers: {total_fertilizer}")
```

## Total Amount of Fertilizer Used in each Village (Groupby)

```python
total_fertilizer_per_village = df.groupby("Village")["Fertilizer Used kg"].sum()
print(f"Total fertilizers used per village: {total_fertilizer_per_village}")
```

## Total Farm Size of All Farmers Combined

```python
sum_all_farm_size = df ["Farm Size (ha)"].sum()
print(f"Total farm size of all farmers combined, in hectares: {sum_all_farm_size}")
```

## Average Farm Size Across all the Farmers

```python
Avg_farm_size = df ["Farm Size (ha)"].mean().round()
print(f"Average farm size of all 50 farmers: {Avg_farm_size}")
```

## Average Yield After Training in Each Village

```python
Avg_yeild_after_training_per_village = df.groupby("Village")["Yield After (kg/ha)"].mean()
print(f"Average yield after training in each village: {Avg_yeild_after_training_per_village}")
```

```python
# creating the chart
Avg_yeild_after_training_per_village.plot(
    kind = "bar",
    color = ["green", "blue", "red"],
    edgecolor = "black"
)

# Adding of labels and title of the bar chart
plt.title("Averger Yeild After Training Per Village", fontsize = 14)
plt.xlabel("Village", fontsize = 12)
plt.ylabel("Average Yeild (kg/ha)", fontsize = 12) 
plt.xticks(rotation = 0)
plt.tight_layout()
plt.show()
```

## Average Amount of Fertilizer Used in Each Village

```python
Avg_fertilizer_per_village = df.groupby("Village")["Fertilizer Used kg"].mean()
print(f"Average fertilizers used per village: {Avg_fertilizer_per_village}")
```

## Median Yield Change Across All Farmers

```python
median_yield_change = df ["Yield Change (%)"].median()
print(f"Middle number of yeild change across all farmers: {median_yield_change}")
```

## Median Fertilizer Used by Farmers Who Applied it Correctly Versus Incorrectly

```python
median_fertilizer_used_correctly_versus_incorrectly = df.groupby("Fertilizer Type")["Yield After (kg/ha)"].median()
print(f"Middle numbers of fertilizer used by farmers who applied it correctly versus incorrectly: {median_fertilizer_used_correctly_versus_incorrectly}")
```

## Median Farm Size for Cassava Farmers Compared to Yam Farmers

```python
median_farm_size_of_cassava_and_yam = df.groupby("Crop Type")["Farm Size (ha)"].median()
print(f"Mid farm size for cassava farmers versus yam farmers: {median_farm_size_of_cassava_and_yam}")
```

## Most Common Crop Grown Among the 50 Farmers

```python
common_crop = df ["Crop Type"].mode()[0]
count = df["Crop Type"].value_counts()[common_crop]
print(f"Most common crops grown by the farmers: {common_crop}")
print(f"Numbers of farmers growing it: {count}")
```

## Village Having the Most Farmers in the Dataset

```python
most_farmers_village = df["Village"].mode()[0]
count = df["Village"].value_counts()[most_farmers_village]
print(f"Village having the most common farmers: {most_farmers_village}")
print(f"Numbers of farmers from the village: {count}")
```

##  Most Common Land Tenure Type

```python
land_tenure_type = df["Land Tenure"].mode()[0]
count = df["Land Tenure"].value_counts()[land_tenure_type]
print(f"Most common land tenure type: {land_tenure_type}")
print(f"Numbers of farmers with the most common land tenure types: {count}")
```

## Is Pest Incidence More Common Than Not Among these Farmers?

```python
pest_incidence = df["Pest Incidence"].mode()[0]
count = df["Pest Incidence"].value_counts()[pest_incidence]
print(f"Was pest incidence common among the farmers? {pest_incidence}")
print(f"How many farmers experienced the incidence: {count}")
```

## Did Most Farmers Apply Fertilizer Correctly or Incorrectly?

```python
fertiizer_application = df["Fertilizer Type"].mode()[0]
count = df["Fertilizer Type"].value_counts()[fertiizer_application]
print(f"Did the farmers applied the fertilizers correctly or incorrectly? {fertiizer_application}")
print(f"How many farmers applied the fertilizer correctly: {count}")
```

## Did Most Farmers Attend the JPIL Training Programme?

```python
training_attendance = df["Training Attended"].mode()[0]
count = df["Training Attended"].value_counts()[training_attendance]
print(f"Did most of the farmers attend the training programme? {training_attendance}")
print(f"Numbers of farmers in attendance: {count}")
```

## Piechart: What Percentage of The 50 Farmers Grow Each Crop Type?

```python
fertilizer_application = df["Crop Type"].value_counts()
print(fertilizer_application)

fertilizer_application.plot (
    kind = "pie",
    colors = ["red", "blue"],
    autopct = "%1.1f%%",
    startangle = 90    
)
plt.title("Percentage of Farmers for each Crop", fontsize = 14)
plt.tight_layout()
plt.show()
plt.savefig("Percentage_crops_of_farmers.png")
```

## How Does the Average Yield Change Percentage Shift as Farm Size Increases from Small to Large Farms?

```python
Avg_yield_change = df.groupby("Farm Size (ha)")["Yield Change (%)"].mean().round(2)

Avg_yield_change.plot(
    kind = "line",
    color = "red",
    marker = "o",
    linewidth = 2
)
plt.title("Average Yield Changes Across Farm Size, fontsize = 14")
plt.xlabel("Farm Size (ha), fontsize = 12")
plt.ylabel("Yield Change (%), fontsize = 12")
plt.tight_layout()
plt.show()
plt.savefig("Trend Between Farm size and Yield Change")
```
