## Total Amount of Fertilizers used by the 50 Farmers

```python
import pandas as pd
df = pd.read_excel ("jpil_farmers_dataset.xlsx", header= 1)
total_fertilizer = df ["Fertilizer Used kg"].sum()
print(f"Total Fertilizer used by the 50 Farmers: {total_fertilizer}")
```

## Total Amount of Fertilizer Used in each Village (Groupby)

```python
total_fertilizer_per_village = df.groupby("Village")["Fertilizer Used kg"].sum()
print(f"Total fertilizers used per village: {total_fertilizer_per_village}")
```

## Total Farm Size of All Farmers Combined

```python
sum_all_farm_size = df ["Farm Size (ha)"].sum()
print(f"Total farm size of all farmers combined, in hectares: {sum_all_farm_size}")
```

## Average Farm Size Across all the Farmers

```python
Avg_farm_size = df ["Farm Size (ha)"].mean().round()
print(f"Average farm size of all 50 farmers: {Avg_farm_size}")
```

## Average Yield After Training in Each Village

```python
Avg_yeild_after_training_per_village = df.groupby("Village")["Yield After (kg/ha)"].mean()
print(f"Average yield after training in each village: {Avg_yeild_after_training_per_village}")
```

## Average Amount of Fertilizer Used in Each Village

```python
Avg_fertilizer_per_village = df.groupby("Village")["Fertilizer Used kg"].mean()
print(f"Average fertilizers used per village: {Avg_fertilizer_per_village}")
```

## Median Yield Change Across All Farmers

```python
median_yield_change = df ["Yield Change (%)"].median()
print(f"Middle number of yeild change across all farmers: {median_yield_change}")
```

## Median Fertilizer Used by Farmers Who Applied it Correctly Versus Incorrectly

```python
median_fertilizer_used_correctly_versus_incorrectly = df.groupby("Fertilizer Type")["Yield After (kg/ha)"].median()
print(f"Middle numbers of fertilizer used by farmers who applied it correctly versus incorrectly: {median_fertilizer_used_correctly_versus_incorrectly}")
```
