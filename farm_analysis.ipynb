## Total Amount of Fertilizers used by the 50 Farmers

```python
import pandas as pd
df = pd.read_excel ("jpil_farmers_dataset.xlsx", header= 1)
total_fertilizer = df ["Fertilizer Used kg"].sum()
print(f"Total Fertilizer used by the 50 Farmers: {total_fertilizer}")
```

## Total Amount of Fertilizer Used in each Village (Groupby)

```python
total_fertilizer_per_village = df.groupby("Village")["Fertilizer Used kg"].sum()
print(f"Total fertilizers used per village: {total_fertilizer_per_village}")
```

## Average Farm Size Across all the Farmers

```python
Avg_farm_size = df ["Farm Size (ha)"].mean().round()
print(f"Average farm size of all 50 farmers: {Avg_farm_size}")
```

## Median Yield Change Across All Farmers

```python
median_yield_change = df ["Yield Change (%)"].median()
print(f"Middle number of yeild change across all fermers: {median_yield_change}")
```
